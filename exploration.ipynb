```python
import csv
import pandas as pd
```

```python
with open('data/d2_input.csv', newline='', ) as csvfile:
    reader = csv.reader(csvfile)
    data = list(reader)
```

```python
#how many SAFE reports
#a report is safe if all numbers are lower than all previous numbers or higher than all previous numbers
#can only differ by between 1 and 3, cannot be equal
```

```python
foo = pd.DataFrame(data, columns=['Entries'])
```

```python
for i in range(len(foo)):
    foo['Entries'].loc[i] = foo['Entries'].loc[i].split()
```

```python

def is_strictly_increasing(lst):
    return all(earlier < later for earlier, later in zip(lst, lst[1:]))

def is_strictly_decreasing(lst):
    return all(earlier > later for earlier, later in zip(lst, lst[1:]))

def varianceisgood(lst):
    return all(1<=y-x<=3 for x,y in zip(lst,lst[1:])) or all(1<=x-y<=3 for x,y in zip(lst,lst[1:]))
    
```

```python
for i in range(len(foo)):
    foo.loc[i, 'GoodVariance'] = varianceisgood(foo.loc[i, 'Entries'])
    foo.loc[i, 'Increasing'] = is_strictly_increasing(foo.loc[i, 'Entries'])
    foo.loc[i, 'Decreasing'] = is_strictly_decreasing(foo.loc[i, 'Entries'])
```

```python
safereports = 0

for i in range(len(foo)):
    if (is_strictly_increasing(foo.loc[i, 'Entries']) and varianceisgood(foo.loc[i, 'Entries'])):
        safereports = safereports +1
    elif (is_strictly_decreasing(foo.loc[i, 'Entries']) and varianceisgood(foo.loc[i, 'Entries'])):
        safereports = safereports +1
print(safereports)

```
